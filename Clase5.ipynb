Sort-Sorted

```python
mi_lista = [5, 1, 4, 2, 8, 3, 6, 7]

# La función sorted() devuelve una NUEVA lista ordenada (usando Timsort)
lista_ordenada_sorted = sorted(mi_lista)
print(f"Lista ordenada con sorted(): {lista_ordenada_sorted}")

# El método list.sort() ordena la lista EN SU LUGAR (usando Timsort)
mi_lista2 = [5, 1, 4, 2, 8, 3, 6, 7]
mi_lista2.sort()
print(f"Lista ordenada con list.sort(): {mi_lista2}")

notas = [50,92,70,100,85]
print("notas originales:", notas)

notas_ordenadas = sorted(notas)
print("notas ordenadas: ", notas_ordenadas)

notas.sort(reverse=True)
print("In place descendente: ", notas)
```

Busqueda Lineal

```python
def busqueda_lineal(lista, objetivo):
    for indice, elemento in enumerate(lista):
        if elemento == objetivo:
            return indice  # Retorna el índice si se encuentra
    return -1  # Retorna -1 si no se encuentra

mi_lista = [3, 7, 1, 9, 4, 6]
objetivo = 5
indice = busqueda_lineal(mi_lista, objetivo)
if indice != -1:
    print(f"El elemento {objetivo} se encontró en el índice {indice}")
else:
    print(f"El elemento {objetivo} no se encontró en la lista")
```

Búsqueda binaria

```python
def busqueda_binaria_iterativa(lista_ordenada, objetivo):
    izquierda = 0
    derecha = len(lista_ordenada) - 1
    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2
        if lista_ordenada[medio] == objetivo:
            return medio
        elif lista_ordenada[medio] < objetivo:
            izquierda = medio + 1
        else:
            derecha = medio - 1
    return -1

mi_lista_ordenada = [1, 3, 4, 6, 7, 9]
objetivo = 5
indice = busqueda_binaria_iterativa(mi_lista_ordenada, objetivo)
if indice != -1:
    print(f"El elemento {objetivo} se encontró en el índice {indice} (búsqueda binaria iterativa)")
else:
    print(f"El elemento {objetivo} no se encontró en la lista (búsqueda binaria iterativa)")
```

```python
def busqueda_binaria_recursiva(lista_ordenada, objetivo, izquierda, derecha):
    if izquierda > derecha:
        return -1
    medio = (izquierda + derecha) // 2
    if lista_ordenada[medio] == objetivo:
        return medio
    elif lista_ordenada[medio] < objetivo:
        return busqueda_binaria_recursiva(lista_ordenada, objetivo, medio + 1, derecha)
    else:
        return busqueda_binaria_recursiva(lista_ordenada, objetivo, izquierda, medio - 1)

mi_lista_ordenada = [1, 3, 4, 6, 7, 9]
objetivo = 5
indice = busqueda_binaria_recursiva(mi_lista_ordenada, objetivo, 0, len(mi_lista_ordenada) - 1)
if indice != -1:
    print(f"El elemento {objetivo} se encontró en el índice {indice} (búsqueda binaria recursiva)")
else:
    print(f"El elemento {objetivo} no se encontró en la lista (búsqueda binaria recursiva)")
```

Búsqueda en diccionarios

```python
mi_diccionario = {"nombre": "Ana", "edad": 30, "ciudad": "México"}
clave_objetivo = "edad"
if clave_objetivo in mi_diccionario:
    valor = mi_diccionario[clave_objetivo]
    print(f"El valor asociado a la clave '{clave_objetivo}' es: {valor}")
else:
    print(f"La clave '{clave_objetivo}' no se encontró en el diccionario")
```

```python
lista_negra_ips = {"10.30.20","192.168.1.5", "172.16.0.8"}

def evaular_peticion(ip_cliente):
    if ip_cliente in lista_negra_ips:
        return "Acceso denegado (403)"
    return "Acceso Granted (200)"

print(evaular_peticion("192.168.1.5"))
```
