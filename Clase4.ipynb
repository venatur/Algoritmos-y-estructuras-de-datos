HEAPs

```python
import heapq

# Crear un min-heap
mi_heap = []
heapq.heappush(mi_heap, 5)
heapq.heappush(mi_heap, 1)
heapq.heappush(mi_heap, 8)
heapq.heappush(mi_heap, 3)

print(f"Heap inicial: {mi_heap}")  # Output: [1, 3, 8, 5] (la representación interna puede variar)

# Obtener el elemento más pequeño
min_elemento = heapq.heappop(mi_heap)
print(f"Elemento mínimo extraído: {min_elemento}, Heap después del pop: {mi_heap}")
# Output: Elemento mínimo extraído: 1, Heap después del pop: [3, 5, 8]

# Insertar un nuevo elemento
heapq.heappush(mi_heap, 2)
print(f"Heap después de insertar 2: {mi_heap}")  # Output: [2, 3, 8, 5]

# Convertir una lista en un heap
mi_lista = [4, 10, 2, 9, 6]
heapq.heapify(mi_lista)
print(f"Lista convertida en heap: {mi_lista}")  # Output: [2, 6, 4, 9, 10]

# Obtener los 2 elementos más grandes
lista_ejemplo = [5, 1, 8, 3, 9, 2]
dos_mayores = heapq.nlargest(2, lista_ejemplo)
print(f"Los 2 elementos más grandes: {dos_mayores}")  # Output: [9, 8]

# Obtener los 3 elementos más pequeños
tres_menores = heapq.nsmallest(3, lista_ejemplo)
print(f"Los 3 elementos más pequeños: {tres_menores}")  # Output: [1, 2, 3]
```

Cola de prioridad

```python
import heapq

class ColaDePrioridad:
    def __init__(self):
        self._heap = []
        self._index = 0  # Para desempatar elementos con la misma prioridad

    def insertar(self, prioridad, elemento):
        # Usamos (-prioridad, index, elemento) para simular una cola de máxima prioridad
        # y para asegurar un orden de extracción determinista en caso de prioridades iguales
        heapq.heappush(self._heap, (-prioridad, self._index, elemento))
        self._index += 1

    def extraer(self):
        if self._heap:
            return heapq.heappop(self._heap)[-1]
        else:
            raise IndexError("La cola de prioridad está vacía")

    def esta_vacia(self):
        return not self._heap

    def peek(self):
        if self._heap:
            return self._heap[0][-1]
        else:
            return None

# Ejemplo de uso de la Cola de Prioridad
cola = ColaDePrioridad()
cola.insertar(3, "Tarea C")
cola.insertar(1, "Tarea A")
cola.insertar(5, "Tarea E")
cola.insertar(2, "Tarea B")
cola.insertar(5, "Tarea F") # Misma prioridad que Tarea E

print("Elementos en la cola (orden de extracción):")
while not cola.esta_vacia():
    print(cola.extraer())
# Output (el orden de elementos con la misma prioridad dependerá del orden de inserción):
# Tarea E
# Tarea F
# Tarea C
# Tarea B
# Tarea A
```

Agoritmo de Dijkstra

```python
import heapq

def dijkstra(grafo, inicio):
    """
    Implementación del algoritmo de Dijkstra para encontrar los caminos más cortos
    desde un nodo de inicio a todos los demás nodos en un grafo ponderado.

    Args:
        grafo (dict): Un diccionario que representa el grafo. Las claves son los nodos
                      y los valores son listas de tuplas (vecino, peso), donde
                      vecino es el nodo adyacente y peso es el peso de la arista.
        inicio (str): El nodo de inicio para encontrar los caminos más cortos.

    Returns:
        tuple: Una tupla que contiene dos diccionarios:
               - distancias: Un diccionario donde las claves son los nodos y los
                             valores son las distancias más cortas desde el nodo
                             de inicio.
               - predecesores: Un diccionario donde las claves son los nodos y los
                               valores son el nodo anterior en el camino más corto
                               desde el nodo de inicio.
    """
    distancias = {nodo: float('inf') for nodo in grafo}
    distancias[inicio] = 0
    cola_prioridad = [(0, inicio)]  # (distancia, nodo)
    predecesores = {nodo: None for nodo in grafo}

    while cola_prioridad:
        dist_actual, nodo_actual = heapq.heappop(cola_prioridad)

        # Ignorar nodos que ya hemos procesado con una distancia menor
        if dist_actual > distancias[nodo_actual]:
            continue

        for vecino, peso in grafo.get(nodo_actual, []):
            distancia = dist_actual + peso
            if distancia < distancias[vecino]:
                distancias[vecino] = distancia
                predecesores[vecino] = nodo_actual
                heapq.heappush(cola_prioridad, (distancia, vecino))

    return distancias, predecesores

def reconstruir_camino(predecesores, destino):
    """
    Reconstruye el camino más corto desde el nodo de inicio hasta el nodo destino
    utilizando el diccionario de predecesores.

    Args:
        predecesores (dict): Un diccionario donde las claves son los nodos y los
                             valores son el nodo anterior en el camino más corto.
        destino (str): El nodo de destino.

    Returns:
        list: Una lista que representa el camino más corto desde el inicio hasta
              el destino (incluyendo el inicio y el destino), o None si no hay
              camino.
    """
    camino = []
    nodo_actual = destino
    while nodo_actual is not None:
        camino.append(nodo_actual)
        nodo_actual = predecesores[nodo_actual]
    return camino[::-1] if camino[0] is not None else None

# Ejemplo de uso:
grafo_ejemplo = {
    'A': [('B', 4), ('C', 2)],
    'B': [('A', 4), ('D', 5), ('E', 1)],
    'C': [('A', 2), ('E', 8)],
    'D': [('B', 5), ('E', 3), ('F', 6)],
    'E': [('B', 1), ('C', 8), ('D', 3), ('F', 7)],
    'F': [('D', 6), ('E', 7)]
}

nodo_inicio = 'A'
nodo_destino = 'F'

distancias, predecesores = dijkstra(grafo_ejemplo, nodo_inicio)

print(f"Distancias más cortas desde {nodo_inicio}: {distancias}")
print(f"Predecesores: {predecesores}")

camino_mas_corto = reconstruir_camino(predecesores, nodo_destino)

if camino_mas_corto:
    print(f"Camino más corto desde {nodo_inicio} a {nodo_destino}: {camino_mas_corto} (Costo: {distancias[nodo_destino]})")
else:
    print(f"No hay camino desde {nodo_inicio} a {nodo_destino}")
```

Algoritmo floyd warshall

```python
import numpy as np

def floyd_warshall(grafo):
    """
    Implementación del algoritmo de Floyd-Warshall para encontrar los caminos
    más cortos entre todos los pares de vértices en un grafo ponderado.

    Args:
        grafo (dict): Un diccionario que representa el grafo. Las claves son los
                      nodos y los valores son otro diccionario donde las claves
                      son los vecinos y los valores son los pesos de las aristas.
                      Por ejemplo: {'A': {'B': 4, 'C': 2}, 'B': {'A': 4, ...}}

    Returns:
        tuple: Una tupla que contiene dos matrices (usando NumPy arrays):
               - distancias: Una matriz donde distancias[i][j] es la distancia
                             más corta entre el vértice i y el vértice j.
               - predecesores: Una matriz donde predecesores[i][j] es el vértice
                               intermedio anterior en el camino más corto de i a j.
                               Si es None, no hay un vértice intermedio directo.
    """
    nodos = list(grafo.keys())
    num_nodos = len(nodos)
    indices = {nodo: i for i, nodo in enumerate(nodos)}

    # Inicializar la matriz de distancias con infinito para todas las parejas
    distancias = np.full((num_nodos, num_nodos), np.inf)
    np.fill_diagonal(distancias, 0)  # La distancia de un nodo a sí mismo es 0

    # Inicializar la matriz de predecesores
    predecesores = np.full((num_nodos, num_nodos), None, dtype=object)

    # Inicializar las distancias con los pesos de las aristas directas
    for u, vecinos in grafo.items():
        for v, peso in vecinos.items():
            distancias[indices[u]][indices[v]] = peso
            predecesores[indices[u]][indices[v]] = u  # El predecesor directo es u

    # Algoritmo de Floyd-Warshall
    for k in range(num_nodos):  # Vértice intermedio
        for i in range(num_nodos):  # Vértice origen
            for j in range(num_nodos):  # Vértice destino
                if distancias[i][k] != np.inf and distancias[k][j] != np.inf:
                    nueva_distancia = distancias[i][k] + distancias[k][j]
                    if nueva_distancia < distancias[i][j]:
                        distancias[i][j] = nueva_distancia
                        predecesores[i][j] = predecesores[k][j]

    return distancias, predecesores, nodos, indices

def reconstruir_camino_floyd_warshall(predecesores, inicio, fin, nodos, indices):
    """
    Reconstruye el camino más corto entre dos nodos utilizando la matriz de
    predecesores generada por el algoritmo de Floyd-Warshall.

    Args:
        predecesores (np.ndarray): La matriz de predecesores.
        inicio (str): El nodo de inicio.
        fin (str): El nodo de destino.
        nodos (list): La lista de todos los nodos en el grafo.
        indices (dict): Un diccionario que mapea nodos a sus índices en las matrices.

    Returns:
        list: Una lista que representa el camino más corto desde el inicio hasta
              el fin, o None si no hay camino.
    """
    if inicio not in indices or fin not in indices:
        return None

    inicio_index = indices[inicio]
    fin_index = indices[fin]

    if predecesores[inicio_index][fin_index] is None:
        if inicio == fin and distancias[inicio_index][fin_index] == 0:
            return [inicio]
        elif distancias[inicio_index][fin_index] == np.inf:
            return None
        else:
            return [inicio, fin]

    camino = [fin]
    actual = fin
    while actual != inicio:
        prev = predecesores[indices[inicio]][indices[actual]]
        if prev is None:
            return None  # No hay camino
        camino.append(prev)
        actual = prev

    return camino[::-1]

# Ejemplo de uso:
grafo_ejemplo = {
    'A': {'B': 4, 'C': 2},
    'B': {'A': 4, 'D': 5, 'E': 1},
    'C': {'A': 2, 'E': 8},
    'D': {'B': 5, 'E': 3, 'F': 6},
    'E': {'B': 1, 'C': 8, 'D': 3, 'F': 7},
    'F': {'D': 6, 'E': 7}
}

distancias, predecesores, nodos, indices = floyd_warshall(grafo_ejemplo)

print("Matriz de Distancias:")
print(distancias)
print("\nMatriz de Predecesores:")
print(predecesores)
print("\nNodos:", nodos)
print("Índices:", indices)

inicio_nodo = 'A'
destino_nodo = 'F'
camino_mas_corto = reconstruir_camino_floyd_warshall(predecesores, inicio_nodo, destino_nodo, nodos, indices)

if camino_mas_corto:
    print(f"\nCamino más corto de {inicio_nodo} a {destino_nodo}: {camino_mas_corto} (Costo: {distancias[indices[inicio_nodo]][indices[destino_nodo]]})")
else:
    print(f"\nNo hay camino de {inicio_nodo} a {destino_nodo}")

inicio_nodo = 'C'
destino_nodo = 'D'
camino_mas_corto = reconstruir_camino_floyd_warshall(predecesores, inicio_nodo, destino_nodo, nodos, indices)

if camino_mas_corto:
    print(f"Camino más corto de {inicio_nodo} a {destino_nodo}: {camino_mas_corto} (Costo: {distancias[indices[inicio_nodo]][indices[destino_nodo]]})")
else:
    print(f"No hay camino de {inicio_nodo} a {destino_nodo}")
```
