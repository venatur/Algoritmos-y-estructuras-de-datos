# Singleton

```python
class Singleton:
    _instancia = None

    def __new__(cls, *args, **kwargs):
        if not cls._instancia:
            cls._instancia = super(Singleton, cls).__new__(cls)
            # Aquí podrías inicializar atributos específicos de la instancia si es necesario
            cls._instancia.valor = None
        return cls._instancia

    def asignar_valor(self, valor):
        self.valor = valor

    def obtener_valor(self):
        return self.valor

# Uso del Singleton
instancia1 = Singleton()
instancia2 = Singleton()

instancia1.asignar_valor(10)
print(f"Valor de instancia1: {instancia1.obtener_valor()}")
print(f"Valor de instancia2: {instancia2.obtener_valor()}")

print(f"¿Son la misma instancia? {instancia1 is instancia2}")
```

# Factory method

```python
from abc import ABC, abstractmethod

class Transport(ABC):
    @abstractmethod
    def deliver(self):
        pass

class Truck(Transport):
    def deliver(self):
        return "Entrega por avion."

class Ship(Transport):
    def deliver(self):
        return "Entrega por barco."

class TransportFactory(ABC):
    @abstractmethod
    def create_transport(self):
        pass

class RoadLogistics(TransportFactory):
    def create_transport(self):
        return Truck()

class SeaLogistics(TransportFactory):
    def create_transport(self):
        return Ship()

def plan_delivery(factory: TransportFactory):
    transport = factory.create_transport()
    return transport.deliver()

road_factory = RoadLogistics()
sea_factory = SeaLogistics()

print(plan_delivery(road_factory))  # Imprime: Entrega por camión.
print(plan_delivery(sea_factory))   # Imprime: Entrega por barco.
```

# Decorator

```python
def bold(func):
    def wrapper(*args, **kwargs):
        return "<b>" + func(*args, **kwargs) + "</b>"
    return wrapper

def italic(func):
    def wrapper(*args, **kwargs):
        return "<i>" + func(*args, **kwargs) + "</i>"
    return wrapper

@italic
@bold
def get_text():
    return "Hola"

print(get_text())  # Imprime: <i><b>Hola</b></i>
```

# Observer

```python
class Subject:
    def __init__(self):
        self._observers = []
        self._state = None

    def attach(self, observer):
        self._observers.append(observer)

    def detach(self, observer):
        self._observers.remove(observer)

    def notify(self):
        for observer in self._observers:
            observer.update(self._state)

    @property
    def state(self):
        return self._state

    @state.setter
    def state(self, new_state):
        self._state = new_state
        self.notify()

class Observer:
    def update(self, state):
        pass

class ConcreteObserverA(Observer):
    def update(self, state):
        print(f"Observador A: El estado ha cambiado a {state}")

class ConcreteObserverB(Observer):
    def update(self, state):
        print(f"Observador B: ¡Recibí una actualización! Nuevo estado: {state}")

sujeto = Subject()
observador_a = ConcreteObserverA()
observador_b = ConcreteObserverB()

sujeto.attach(observador_a)
sujeto.attach(observador_b)

sujeto.state = "Nuevo estado 1"
sujeto.state = "Ot "
```
